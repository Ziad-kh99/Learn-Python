# Sockets and Networking Programming:

- Socket is the endpoint of the network connection between two devices, usually client and server.

## Socket Protocol:
- The socket either use TCP (reliable) or UDP (Faster) protocols to establish the connection.
1. TCP Socket (socket.SOCK_STREAM)
2. UDP Socket (socket.SOCK_DGRAM)

## Socket Type:
- Also there're two types of sockets we can establish:
    1. Internet Socket. (socket.AF_INET)
    2. Unix Socket.

```python
# Server Side:

import socket

# Create TCP-Internet Scoket
s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)   

# Bind the socket with IP-Port
s.bind(('127.0.0.1', 55555))

# Open the socket to accept connections
s.listen()

while(True):
    client, address = s.accept()
    print(f'connected to {address}')
    client.send('You\'r connected'.encode())
    client.close() 
```

```python
# Client Side:
import socket

# Create Internet-TCP socket
s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Connect to socket in server-side
s.connect(('127.0.0.1', 55551))
message = s.recv(1024)
s.close()

print(message.decode())
```
